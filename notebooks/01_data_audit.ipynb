```python
import os
import pandas as pd
import json

# Paths
DATA_DIR = '../data/dataset'
print('Data directory contents:')
print('Train:', os.listdir(os.path.join(DATA_DIR, 'train')))
print('Test:', os.listdir(os.path.join(DATA_DIR, 'test')))
```

### 2. Inspect Ground Truth Matching Distribution

```python
# Sample ground truth
gt_df = pd.read_csv(os.path.join(DATA_DIR, 'train', 'train_ground_truth.tsv'), sep='\t', nrows=50000, keep_default_na=False)
gt_df['match_count'] = gt_df['matched_entity_ids'].apply(lambda x: 0 if not x.strip() else len(x.split(',')))

print('Match count distribution in first 50,000 entities:')
print(gt_df['match_count'].value_counts().sort_index())
```

### 3. Inspect Noise Patterns (Examples from Audit)

```python
with open('../reports/sample_pairs.json', 'r', encoding='utf-8') as f:
    sample_data = json.load(f)

print('--- SINGLETON EXAMPLES (No matches) ---')
for s in sample_data['singletons'][:3]:
    print(f"ID: {s['id']} | Name: {s['details']['business_name']} | Country: {s['details']['country']}")
    print(f"Address: {s['details']['business_address']}\n")

print('--- NOISY MATCH EXAMPLES ---')
for p in sample_data['pairs'][:8]:
    print(f"Pair: {p['s1_id']} <-> {p['target_id']} ({p['country']})")
    print(f"  S1 Name    : {p['s1_name']}")
    print(f"  Target Name: {p['target_name']}")
    print(f"  S1 Addr    : {p['s1_address']}")
    print(f"  Target Addr: {p['target_address']}")
    print(f"  Name Exact?: {p['name_exact_case_insensitive']} | Addr Empty?: {p['target_address_empty']}\n")
```
